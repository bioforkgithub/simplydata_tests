# 00 Summary of all comparisons

Every number compared in the other notebooks, read from `results/` (each notebook saves its own table there when it
runs). Columns: the dataset, the simplydata word (method), the quantity, simplydata's value, the reference value and
where it comes from, the difference, the tolerance and what the comparison found.

**Only simplydata runs in this notebook**: each results file is read with `sdata.table`, the files are stacked with
`add_rows_from`, counted with `count_rows(for_each_group_in=...)` and laid out with `turn_rows_into_columns`.

What each comparison found (the column "result", written by each notebook):
- **agrees**: within the stated tolerance.
- **simplydata error – being fixed**: a simplydata bug (FINDINGS.md class A), until its fix is released.
- **explained difference – reference**: the reference rounds, stops early, uses another definition or has its own
  known bug (class C).
- **explained difference – machine**: the last digits differ between CPUs or systems.
- **explained difference – precision limit**: the limit of double precision or of the package simplydata calls (B).
- **difference not yet explained**: outside the tolerance and not yet in `results/known_fails.csv`; CI fails on it.
- **waiting for word**: the check is written, but simplydata has no word yet to give the number back.
- **skipped**: the comparison could not be made (the reason is given).

Every difference keeps its finding id, a one-line explanation and a link to [FINDINGS.md](../FINDINGS.md). The machine
column "status" keeps PASS, FAIL and SKIP, so nothing is hidden. Notebooks with a reference file name the kind of each
reference (printed, certified, expected or run reference) in the column "kind".

Run the other notebooks first (or `python scripts/run_notebooks.py`, which runs this one last).

```python
import simplydata as sdata

from checks import RESULTS, quiet, result_files, versions

versions("simplydata")
KEY = ["notebook", "dataset", "method", "quantity", "reference source"]
WORDS = ["kind", "result", "finding", "explanation", "link"]
COLUMNS = KEY + ["simplydata", "reference", "difference", "tolerance", "tolerance type", "score", "status", "reason"] + WORDS

table = None
for path in result_files():
    part = quiet(sdata.table, path, separator=",")
    names = part.get_column_names()
    for column in WORDS:
        if column not in names:
            quiet(part.add_column, column, value="")
    quiet(part.keep_columns, *COLUMNS)
    table = part if table is None else quiet(table.add_rows_from, part)
quiet(table.save, str(RESULTS / "all_comparisons.csv"))
print(f"{quiet(table.count_rows)} comparisons from {len(result_files())} notebooks; saved to results/all_comparisons.csv")


def count_by(*columns, across="result"):
    """How many comparisons, one row per value of `columns`, one column per value of `across`."""
    counted = quiet(table.count_rows, for_each_group_in=[*columns, across])
    return quiet(counted.turn_rows_into_columns, names_from=across, values_from="count")
```

## What each comparison found, by notebook

```python
count_by("notebook").show()
quiet(table.count_rows, for_each_group_in="result").show()
```

## The machine column (PASS, FAIL, SKIP), by notebook

```python
count_by("notebook", across="status").show()
```

## By kind of reference

Printed or certified values (NIST, the printed papers, tables, manuals and help pages, values exact by construction),
what simplydata must say of its own answer ("expected"), and a program run once by the notebook's reference script
("run reference": the parent package or R). Notebooks still in the older layout leave the column empty.

```python
count_by("kind").show()
```

## By dataset and word

```python
count_by("dataset", "method").show()
```

## Every difference, with its explanation

`score` is the LRE (significant digits that agree) for NIST and other `lre` comparisons, otherwise the absolute or
relative difference. "finding" is the FINDINGS.md section and "link" goes to it; "difference not yet explained" is
outside the tolerance and not yet in `results/known_fails.csv` (CI fails on it).

```python
quiet(table.keep_rows_where, "status", is_equal_to="FAIL")
quiet(table.count_rows, for_each_group_in="finding").show()
quiet(table.keep_rows_where, "result", is_equal_to="difference not yet explained")
print(f"{quiet(table.count_rows)} differences not yet explained")
quiet(table.undo)
quiet(table.keep_columns, "notebook", "dataset", "method", "quantity", "simplydata", "reference", "reference source",
      "score", "tolerance", "tolerance type", "result", "finding", "explanation", "link").show()
quiet(table.undo)
quiet(table.undo);
```

## Every SKIP, and the checks waiting for a simplydata word

```python
quiet(table.keep_rows_where, "status", is_equal_to="SKIP")
quiet(table.keep_rows_where, "reason", starts_with="waiting for word")
print(f"{quiet(table.count_rows)} comparisons wait for a simplydata word")
quiet(table.undo)
quiet(table.keep_columns, "notebook", "dataset", "method", "quantity", "reference source", "result", "reason").show()
quiet(table.undo)
quiet(table.undo);
```
