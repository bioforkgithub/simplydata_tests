```python
import simplydata as sdata

from reference_checks import Comparisons, Result, quiet

cmp = Comparisons("32_models_planning_more")
DATA = "../data/"


def one(result):
    return Result(result).row(1)
```

## 1. Comparing models for ordered groups, groups and shares; model averages for counts with many zeros and positive numbers

```python
housing = sdata.table(cmp.data_file("housing_one_row_per_tenant"))
for outcome in ("ordered groups", "groups"):
    c = Result(quiet(housing.compare_models, "Sat", using=["Infl", "Type"], against=["Infl", "Type", "Cont"], outcome=outcome,
                     in_order=["Low", "Medium", "High"] if outcome == "ordered groups" else None))
    cmp.check_all("housing (MASS)", f'compare_models (outcome="{outcome}")', {"statistic": c.cell(3, "statistic"), "p-value": c.cell(3, "p-value")})
gas = sdata.table(cmp.data_file("GasolineYield"))
c = Result(quiet(gas.compare_models, "yield", using=["temp"], against=["batch", "temp"], outcome="shares"))
cmp.check_all("GasolineYield (betareg)", 'compare_models (outcome="shares")', {"statistic": c.cell(3, "statistic"), "p-value": c.cell(3, "p-value")})
z = Result(quiet(sdata.table(cmp.data_file("bioChemists")).find_model_averages_of, "art", for_each_group_in="mar", using=["mar", "ment"],
                 outcome="counts with many zeros"), key="term")
cmp.check_all("bioChemists (pscl)", 'find_model_averages_of (outcome="counts with many zeros")',
              {f"{g} {q}": z[g, c] for g in ("Married", "Single") for q, c in (("model average", "model average"), ("standard error", "standard error"))})
clotting = sdata.table(cmp.data_file("clotting_two_lots"))
quiet(clotting.add_column, "log_u", "log(u)")
g = Result(quiet(clotting.find_model_averages_of, "time", for_each_group_in="lot", using=["lot", "log_u"], outcome="positive numbers"), key="term")
cmp.check_all("clotting times, two lots (R glm help page)", 'find_model_averages_of (outcome="positive numbers")',
              {f"{lot} {q}": g[lot, c] for lot in ("lot 1", "lot 2") for q, c in (("model average", "model average"), ("standard error", "standard error"))})
```

## 4. Power: paired and chi-square tests (pwr vignette, printed), two percents (R power.prop.test), your own distributions

`drawing_from=` two log-normal distributions (log-scale averages 0 and 0.4, spread 0.5): simulated power of Welch's
t-test against SciPy `stats.power` with the same seed.

```python
P = "planning settings"
cmp.check(P, 'find_chance_of_finding (for_test="compare_before_after", effect_size=0.32, values_per_group=24, sides=1)', "power",
          one(quiet(sdata.find_chance_of_finding, for_test="compare_before_after", effect_size=0.32, values_per_group=24, sides=1))["power"])
paired = one(quiet(sdata.find_how_many_needed, for_test="compare_before_after", effect_size=0.32, sides=1))
cmp.check(P, 'find_how_many_needed (for_test="compare_before_after", effect_size=0.32, sides=1)', "pairs (exact)", paired[[k for k in paired if "exact" in k][0]])
# pwr's N is the number of observations in all: 4 groups of 25 (3 degrees of freedom) and 2 groups of 50 (1 degree of freedom)
cmp.check(P, 'find_chance_of_finding (for_test="compare_counts", effect_size=0.2886751, values_per_group=25, groups=(4, 2))', "power",
          one(quiet(sdata.find_chance_of_finding, for_test="compare_counts", effect_size=0.2886751, values_per_group=25, groups=(4, 2)))["power"])
counts = one(quiet(sdata.find_how_many_needed, for_test="compare_counts", effect_size=0.2886751, groups=(4, 2)))
cmp.check(P, 'find_how_many_needed (for_test="compare_counts", effect_size=0.2886751, groups=(4, 2))', "values in all (exact)",
          4 * counts["values per group (exact)"])
cmp.check(P, 'find_chance_of_finding (for_test="compare_counts", effect_size=0.2182179, values_per_group=50, groups=(2, 2), significance=0.01)', "power",
          one(quiet(sdata.find_chance_of_finding, for_test="compare_counts", effect_size=0.2182179, values_per_group=50, groups=(2, 2), significance=0.01))["power"])
percents = one(quiet(sdata.find_how_many_needed, for_test="compare_counts", percents=[30, 15]))
cmp.check(P, 'find_how_many_needed (for_test="compare_counts", percents=[30, 15])', "values per group (exact)", percents[[k for k in percents if "exact" in k][0]])
cmp.check(P, 'find_chance_of_finding (for_test="compare_counts", percents=[30, 15], values_per_group=120)', "power",
          one(quiet(sdata.find_chance_of_finding, for_test="compare_counts", percents=[30, 15], values_per_group=120))["power"])
first = sdata.distribution("lognormal", log_average=0, log_standard_deviation=0.5)
second = sdata.distribution("lognormal", log_average=0.4, log_standard_deviation=0.5)
own = Result(quiet(sdata.find_chance_of_finding, drawing_from=[first, second], values_per_group=[20, 40], tries=2000, with_seed=1))
for i, n in enumerate((20, 40), start=1):
    cmp.check(P, "find_chance_of_finding (drawing_from=[two lognormal distributions], values_per_group=[20, 40], tries=2000, with_seed=1)",
              f"power with {n} per group", own.cell(i, "power"))
```

## 5. A confirmatory factor model with empty cells kept (psych's bfi; lavaan full-information maximum likelihood)

```python
bfi = sdata.table(cmp.data_file("bfi"))
cfa = Result(quiet(bfi.check_hidden_factors, measured_by={"agree": ["A1", "A2", "A3", "A4", "A5"], "consc": ["C1", "C2", "C3", "C4", "C5"]}), key="term")
chi = [t for t in cfa.index if "chi-square" in t.lower()][0]
ours = {"chi-square": cfa[chi, "estimate"], "degrees of freedom": cfa[chi, "degrees of freedom"]}
for name in ("CFI", "RMSEA"):
    found = [t for t in cfa.index if t.upper().startswith(name)]
    if found:
        ours[name] = cfa[found[0], "estimate"]
for factor, items in (("agree", ["A1", "A2", "A3", "A4", "A5"]), ("consc", ["C1", "C2", "C3", "C4", "C5"])):
    for item in items:
        ours[f"standardised loading of {item}"] = cfa[f"{factor} measured by {item}", "estimate"]
cmp.check_all("bfi (psych)", "check_hidden_factors (measured_by=), empty cells kept (FIML)", ours)
```

## 6. Other kinds of design table (R agricolae with the same seed)

```python
for kind, levels in (("completely random", {"treatment": ["x", "y", "z"]}), ("randomised blocks", {"treatment": ["x", "y", "z"]}),
                     ("latin square", {"treatment": ["x", "y", "z", "w"]})):
    for seed in (2, 9):
        options = {"repeats": 4} if kind != "latin square" else {}
        d = Result(quiet(sdata.make_design_table, levels, of_kind=kind, with_seed=seed, **options))
        check = f'make_design_table (of_kind="{kind}", repeats=4, with_seed={seed})' if kind != "latin square" else f'make_design_table (of_kind="latin square", with_seed={seed})'
        cmp.check_vector("designs (same seed)", check, "f1", d["treatment"])
        if kind == "randomised blocks":
            cmp.check_vector("designs (same seed)", check, "block", d["block"])
        if kind == "latin square":
            cmp.check_vector("designs (same seed)", check, "row", d["row"])
            cmp.check_vector("designs (same seed)", check, "col", d["column"])
```

## 7. A Bayesian count model with a seed (rstanarm `stan_glm`, the same seed): warpbreaks

```python
bayes = Result(quiet(sdata.table(DATA + "warpbreaks.csv").explain, "breaks", using=["wool", "tension"], outcome="counts", test_method="bayesian",
                     with_seed=1), key="term")
ratio = "effect"   # for counts, the effect column holds the rate ratio (the posterior median of exp(coefficient))
terms = ["(intercept)", "wool (B)", "tension (L)", "tension (M)"]
logs = list(sdata.numbers([bayes[t, ratio] for t in terms]).take_log())
cmp.check_all("warpbreaks", 'explain (outcome="counts", test_method="bayesian", with_seed=1)',
              {f"{t} posterior median (log scale)": v for t, v in zip(terms, logs)})
```

## 8. The Bayesian model checks: `check_model_of(test_method="bayesian")`

`show="numbers"` gives back R-hat and the effective draws of each parameter, the divergent transitions and the
posterior predictive check (the count of each value in the data against the median and 90 % interval of its count in
50 made-up sets). `show="prior check"` draws the prior predictive check, read back with `chart.get_numbers_drawn()`
(bars: the data; dots: the median; error bars: the 90 % interval). Compared with R rstanarm run with the same calls
and the same seed (`posterior_predict(draws = 50, seed = 1)`, `posterior::summarise_draws`); for the prior check
every k-th value, so that about 60 are compared, and the number of bars.

```python
warp = sdata.table(DATA + "warpbreaks.csv")
numbers = Result(quiet(warp.check_model_of, "breaks", using=["wool", "tension"], outcome="counts", test_method="bayesian",
                       show="numbers"))
check = 'check_model_of (test_method="bayesian", show="numbers")'
for name in ("R-hat", "effective draws in the middle (bulk)", "effective draws in the tails"):
    cmp.check_vector("warpbreaks", check, name, Result(numbers.table, where={"check": name})["value"])
cmp.check("warpbreaks", check, "divergent transitions", Result(numbers.table, where={"check": "divergent transitions"})["value"][0])
data = Result(numbers.table, where={"term": "count in the data"})
made = Result(numbers.table, where={"term": "count in the made-up sets (median, 90 % interval)"})
for name, values in (("value", data["at"]), ("count in the data", data["value"]), ("median count in the made-up sets", made["value"]),
                     ("5 % of the made-up sets", made["low"]), ("95 % of the made-up sets", made["high"])):
    cmp.check_vector("warpbreaks", check, f"posterior predictive check: {name}", values)

chart = quiet(warp.check_model_of, "breaks", using=["wool", "tension"], outcome="counts", test_method="bayesian",
              show="prior check")
drawn = quiet(chart.get_numbers_drawn)
check = 'check_model_of (test_method="bayesian", show="prior check")'
bars = Result(drawn, where={"kind": "bar"})
dots = Result(drawn, where={"kind": "dot"})
error_bars = Result(drawn, where={"kind": "error bar"})
cmp.check("warpbreaks", check, "drawn bars (number)", len(bars))
for name, values in (("drawn bars: value", bars["x"]), ("drawn bars: count in the data", bars["y"]),
                     ("drawn dots: median count in the made-up sets", dots["y"]),
                     ("drawn error bars: 5 % of the made-up sets", error_bars["low"]),
                     ("drawn error bars: 95 % of the made-up sets", error_bars["high"])):
    cmp.check_vector("warpbreaks", check, name, values)
```
