# 20 Tests of counts and percents, pairs of groups, matched columns, adjusted p-values and correlations

`table.compare_counts_of`, `sdata.compare_counts`, `table.compare_counts_to`, `table.compare_percent_to`,
`table.compare_each_pair_of`, `table.compare_matched_columns`, `table.add_adjusted_p_value_column` and
`table.compare_correlations`, and how simplydata answers the misspelt option values the gap list names.

**Only simplydata runs in this notebook.** The reference values are computed by
`scripts/references/20_tests_counts_pairs_matched.py` and stored in `references/20_tests_counts_pairs_matched.csv`
with source, kind and link: the numbers printed in the SciPy docstrings of `tukey_hsd`, `dunnett` and
`page_trend_test`; R `fisher.test`, `chisq.test`, `prop.trend.test`, `binom.test`, `friedman.test`, `quade.test`,
`p.adjust`, `TukeyHSD`, `pairwise.t.test`, `pairwise.wilcox.test`, DescTools, multcomp, PMCMRplus, qvalue, crank
and R Exact; and formulas written out (Barnard's test over a grid, the Cochran-Armitage statistic, Cohen's h,
Dunn's test, the Scheffé test, Fisher's and Steiger's z).

Data: the examples of R's `fisher.test` (Fisher's tea tasting; Fisher's twins and convictions; Agresti's job
satisfaction), `prop.trend.test` (Fleiss's smokers), `binom.test` (Conover's peas) and `friedman.test`
(Hollander & Wolfe's rounding first base) help pages, DescTools `CochranQTest`, the SciPy docstrings above,
HairEyeColor, PlantGrowth, InsectSprays, the Hedenfalk p-values (`data/`) and mtcars.

```python
import simplydata as sdata

from checks import ReferenceChecks, data_file, quiet, read_numbers, read_row, read_rows, versions

NOTEBOOK = "20_tests_counts_pairs_matched"
versions("simplydata")
checks = ReferenceChecks(NOTEBOOK)


def load(name):
    return quiet(sdata.table, data_file(NOTEBOOK, name), separator=",")


def answer(word, *args, **options):
    return read_row(quiet(word, *args, **options))


def counts_of(table, rows_column, columns_column):
    """The table of counts (one list per row), read from count_rows(across=) with simplydata."""
    found = read_rows(quiet(table.count_rows, for_each_group_in=rows_column, across=columns_column))
    names = [c for c in found[0] if c != rows_column]
    return [[row[c] for c in names] for row in found]
```

## `compare_counts_of` and `sdata.compare_counts`: R's `fisher.test` help page

Tea tasting and the twins' convictions (2 x 2): Fisher's exact test, Barnard's and Boschloo's tests and the
chi-square test, from rows and from counts. The job satisfaction table (4 x 4): Fisher's test from 100 000 random
tables (`with_seed=1`) and exact (`p_value_by="exact"`), the G-test, the standardised residuals, and the shuffled
chi-square p-value.

```python
for name in ("TeaTasting", "Convictions"):
    t = load(f"{name}_rows")
    ds = f"{name} (R fisher.test example)"
    counts = counts_of(t, "row", "column")
    for test in ("fisher", "barnard", "boschloo"):
        checks.check(ds, f'compare_counts_of (test_method="{test}")', "p-value",
                     answer(t.compare_counts_of, "column", for_each_group_in="row", test_method=test)["p-value"])
        checks.check(ds, f'sdata.compare_counts (test_method="{test}")', "p-value",
                     answer(sdata.compare_counts, counts, test_method=test)["p-value"])
    checks.check_values(ds, "sdata.compare_counts", answer(sdata.compare_counts, counts))
```

```python
jobs = load("job")
ds = "job satisfaction (R fisher.test example; Agresti 2002)"
found = answer(jobs.compare_counts_of, "satisfaction", for_each_group_in="income", test_method="fisher", with_seed=1)
checks.check(ds, 'compare_counts_of (test_method="fisher", 4 x 4, with_seed=1)', "p-value", found["p-value"])
found = answer(jobs.compare_counts_of, "satisfaction", for_each_group_in="income", test_method="fisher", p_value_by="exact")
checks.check(ds, 'compare_counts_of (test_method="fisher", p_value_by="exact")', "p-value", found["p-value"])
checks.check_values(ds, 'compare_counts_of (test_method="g")',
                    answer(jobs.compare_counts_of, "satisfaction", for_each_group_in="income", test_method="g"))
cells = read_rows(quiet(jobs.compare_counts_of, "satisfaction", for_each_group_in="income", show="cells"))
checks.check_values(ds, 'compare_counts_of (show="cells")',
                    {f"standardised residual {row['income']}, {row['satisfaction']}": row["standardised residual"] for row in cells})
found = answer(jobs.compare_counts_of, "satisfaction", for_each_group_in="income", p_value_by="shuffling", shuffles=9999, with_seed=1)
checks.check(ds, 'compare_counts_of (p_value_by="shuffling", shuffles=9999, with_seed=1)', "p-value", found["p-value"])
```

Fleiss's smokers by age group (Cochran-Armitage trend), the eye colours of HairEyeColor (goodness of fit) and
Conover's peas (the exact binomial test, one-sided forms and three intervals).

```python
checks.check_values("smokers (R prop.trend.test example; Fleiss 1981)", 'compare_counts_of (test_method="trend", in_order=)',
                    answer(load("smoke").compare_counts_of, "smoking", for_each_group_in="age group", test_method="trend",
                           in_order=["g1", "g2", "g3", "g4"]))
eyes = load("eyes")
ds = "HairEyeColor eye colours (Snee 1974)"
checks.check_values(ds, "compare_counts_to", answer(eyes.compare_counts_to, "eye"))
checks.check_values(ds, "compare_counts_to (expected=)",
                    answer(eyes.compare_counts_to, "eye", expected={"Brown": 0.4, "Blue": 0.3, "Hazel": 0.15, "Green": 0.15}))
peas = load("pea_rows")
ds = "Mendel's peas (R binom.test example; Conover 1971)"
found = answer(peas.compare_percent_to, "plant", value="giant", percent=75)
checks.check_values(ds, "compare_percent_to", {**found, "effect size (Cohen's h)": found["effect size"]})
for options in [{"sides": 1, "higher": "percent"}, {"sides": 1, "lower": "percent"}, {"alternative": "less"}]:
    checks.check(ds, "compare_percent_to (" + ", ".join(f"{k}={v!r}" for k, v in options.items()) + ")", "p-value",
                 answer(peas.compare_percent_to, "plant", value="giant", percent=75, **options)["p-value"])
for method in ("wilson", "agresti-coull"):
    checks.check_values(ds, f'compare_percent_to (interval_by="{method}")',
                        answer(peas.compare_percent_to, "plant", value="giant", percent=75, interval_by=method))
```

## `compare_each_pair_of`

The `tukey_hsd` docstring (its printout gives 3 decimals) with R `TukeyHSD`; the `dunnett` docstring (Dunnett
1955) with R multcomp; then every follow-up test on PlantGrowth and InsectSprays, and the letters.

```python
def pair_values(found, prefix, column):
    """Each pair's number under both orders of its two groups ("a - b" and "b - a")."""
    values = {}
    for row in found:
        a, b = row["group 1"], row["group 2"]
        values[f"{prefix} {a} - {b}"] = row[column]
        values[f"{prefix} {b} - {a}"] = row[column]
    return values


found = read_rows(quiet(load("tk").compare_each_pair_of, "y", for_each_group_in="group", test_method="tukey"))
pcol = [c for c in found[0] if c.startswith("p-value")][0]
values = {}
for row in found:
    i, j = row["group 1"][-1], row["group 2"][-1]
    for quantity in ("difference", "difference 95 % CI low", "difference 95 % CI high"):
        values[f"{quantity} {i} - {j}"] = row[quantity]
    values[f"p-value {i} - {j}"] = row[pcol]
    a, b = row["group 1"], row["group 2"]
    values[f"p-value {b}-{a}"] = values[f"p-value {a}-{b}"] = row[pcol]
    values[f"difference {b}-{a}"] = -row["difference"]   # R's "group1-group0" is group1 minus group0
    values[f"difference {a}-{b}"] = row["difference"]
checks.check_values("SciPy tukey_hsd docstring", 'compare_each_pair_of (test_method="tukey")', values)

found = read_rows(quiet(load("dn").compare_each_pair_of, "y", for_each_group_in="group", against="control", with_seed=1))
values = {}
for row in found:
    values[f"statistic {row['group 1']}"] = row["statistic"]
    values[f"p-value {row['group 1']}"] = row["p-value (adjusted)"]
for method in ("compare_each_pair_of (against=, with_seed=1)", "compare_each_pair_of (against=)"):
    checks.check_values("SciPy dunnett docstring (Dunnett 1955)", method, values)
```

```python
plants, insects = load("plants"), load("insects")
for name, table, column, group in [("PlantGrowth (Dobson 1983)", plants, "weight", "group"),
                                   ("InsectSprays (Beall 1942)", insects, "count", "spray")]:
    for test in ("welch", "mann-whitney", "games-howell", "lsd", "dunn", "conover", "scheffe"):
        options = {"based_on": "ranks"} if test == "mann-whitney" else {"test_method": test}
        found = read_rows(quiet(table.compare_each_pair_of, column, for_each_group_in=group, **options))
        pcol = [c for c in found[0] if c.startswith("p-value (") or c == "p-value (adjusted)"]
        option = 'based_on="ranks"' if test == "mann-whitney" else f'test_method="{test}"'
        checks.check_values(name, f"compare_each_pair_of ({option})", pair_values(found, "adjusted p-value", pcol[0] if pcol else "p-value"))
    letters = read_rows(quiet(table.compare_each_pair_of, column, for_each_group_in=group, show="letters"))
    share = {}
    for la in letters:
        for lb in letters:
            share[f"{la['group']} and {lb['group']} share a letter"] = float(bool(set(la["letters"]) & set(lb["letters"])))
    checks.check_values(name, 'compare_each_pair_of (show="letters")', share)
```

## `compare_matched_columns`, `add_adjusted_p_value_column` and `compare_correlations`

```python
rounding = load("rounding")
methods = rounding.get_column_names()
ds = "rounding first base (R friedman.test example; Hollander & Wolfe 1973)"
found = answer(rounding.compare_matched_columns, *methods)
checks.check_values(ds, "compare_matched_columns", {**found, "effect size (Kendall's W)": found["effect size"]})
checks.check_values(ds, 'compare_matched_columns (test_method="quade")', answer(rounding.compare_matched_columns, *methods, test_method="quade"))
found = answer(load("page").compare_matched_columns, "first", "second", "third", test_method="page")
checks.check_values("SciPy page_trend_test docstring (Page 1963)", 'compare_matched_columns (test_method="page")',
                    {"statistic (L)": found["statistic"], "p-value": found["p-value"]})
checks.check_values("SAS FREQ example (R DescTools CochranQTest example)", "compare_matched_columns (yes/no columns: Cochran's Q)",
                    answer(load("answers").compare_matched_columns, "A", "B", "C"))

ds = "Hedenfalk et al. (2001) p-values"
for method, size in [("holm", 3170), ("bonferroni", 3170), ("benjamini-hochberg", 3170), ("bh", 3170), ("fdr", 3170),
                     ("benjamini-yekutieli", 3170), ("hochberg", 3170), ("hommel", 300)]:
    t = quiet(sdata.table, "../data/fetched/hedenfalk_p.csv", separator=",") if size == 3170 else load("hedenfalk_300")
    source = "p" if size == 3170 else "p-value"
    quiet(t.add_adjusted_p_value_column, "adjusted", from_column=source, test_method=method)
    checks.check(ds, f'add_adjusted_p_value_column (test_method="{method}")', f"adjusted p-values ({size} rows)",
                 read_numbers(quiet(t.get_column, "adjusted")))
for method, column in [("storey", "q"), ("q-values", "q"), ("local fdr", "lfdr"), ("lfdr", "lfdr")]:
    t = quiet(sdata.table, "../data/fetched/hedenfalk_p.csv", separator=",")
    quiet(t.add_adjusted_p_value_column, from_column="p", test_method=method)
    new = t.get_column_names()[-1]
    checks.check(ds, f'add_adjusted_p_value_column (test_method="{method}")', f"{column} (3170 rows)",
                 read_numbers(quiet(t.get_column, new)))

cars = load("mtcars")
checks.check_values("mtcars mpg, wt by transmission", "compare_correlations (for_each_group_in=)",
                    answer(cars.compare_correlations, "mpg", "wt", for_each_group_in="transmission"))
checks.check_values("mtcars mpg with wt and with hp", "compare_correlations (two overlapping pairs)",
                    answer(cars.compare_correlations, ["mpg", "wt"], ["mpg", "hp"]))
```

## Misspelt option values

simplydata does not guess: it stops and names the right value. Each is checked to stop with the right suggestion
(1 = yes).

```python
cases = [("compare_each_pair_of", lambda: plants.compare_each_pair_of("weight", for_each_group_in="group", test_method="turkey"), "tukey"),
         ("add_adjusted_p_value_column", lambda: load("hedenfalk_300").add_adjusted_p_value_column(test_method="hollm"), "holm"),
         ("find_percent_of_rows_where", lambda: plants.find_percent_of_rows_where("weight", is_above=5, interval_by="wilsn"), "wilson")]
for word, call, right in cases:
    try:
        quiet(call)
        said = ""
    except Exception as problem:  # noqa: BLE001
        said = str(problem)
    checks.check("PlantGrowth / Hedenfalk", f"{word} (misspelt value)", f'stops and suggests "{right}"',
                 float(f'"{right}"' in said and "Did you mean" in said))
```
